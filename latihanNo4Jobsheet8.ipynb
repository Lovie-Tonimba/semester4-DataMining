```python
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
# menggunakan fitur Sepal Length dan Sepal Width
X = iris.data[:, [0, 1]]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# 3. enggunakan Agglomerative Clustering
agg_clust = AgglomerativeClustering(n_clusters=4)
agg_labels = agg_clust.fit_predict(X_scaled)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=agg_labels, cmap='viridis', s=100)
plt.title("Hasil Agglomerative Clustering (Data Iris)")
plt.xlabel("Sepal Length (Standardized)")
plt.ylabel("Sepal Width (Standardized)")
plt.grid(True)
plt.show()
```
